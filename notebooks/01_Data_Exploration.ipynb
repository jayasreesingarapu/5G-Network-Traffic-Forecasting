```python
import pandas as pd
import time

file_path = "../data1.csv/data1.csv"

total_rows = 0
missing_counts = None

unique_grids = set()
unique_timestamps = set()

min_timestamp = None
max_timestamp = None

start = time.time()

for i, chunk in enumerate(
    pd.read_csv(file_path, chunksize=100000)
):
    total_rows += len(chunk)

    # Missing values
    current_missing = chunk.isnull().sum()

    if missing_counts is None:
        missing_counts = current_missing
    else:
        missing_counts += current_missing

    # Unique geographical regions
    unique_grids.update(chunk["GridID"].unique())

    # Unique timestamps
    unique_timestamps.update(chunk["TimeInterval"].unique())

    # Timestamp range
    current_min = chunk["TimeInterval"].min()
    current_max = chunk["TimeInterval"].max()

    min_timestamp = (
        current_min if min_timestamp is None
        else min(min_timestamp, current_min)
    )

    max_timestamp = (
        current_max if max_timestamp is None
        else max(max_timestamp, current_max)
    )

    if (i + 1) % 10 == 0:
        print(
            f"Processed {total_rows:,} rows "
            f"in {round(time.time() - start, 1)} seconds",
            flush=True
        )

print("\n=== DATASET SUMMARY ===")

print("Total rows:", total_rows)
print("Unique GridIDs:", len(unique_grids))
print("Unique timestamps:", len(unique_timestamps))

print("\nStart date:",
      pd.to_datetime(min_timestamp, unit="ms", utc=True))

print("End date:",
      pd.to_datetime(max_timestamp, unit="ms", utc=True))

print("\nMissing values:")
print(missing_counts)

print("\nMissing percentages:")
print((missing_counts / total_rows * 100).round(2))

print("\nTotal time:",
      round(time.time() - start, 2), "seconds")
```
